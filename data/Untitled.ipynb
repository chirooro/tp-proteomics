```python
import pandas as pd
import numpy as np

df = pd.read_csv('TCL_wt1.tsv', sep="\t",
    dtype={
        'Accession': str, 
        'Description': str,
        'Gene Symbol': str,
        'Corrected Abundance ratio (1.53)': float,
        'Log2 Corrected Abundance Ratio': float,
        'Abundance Ratio Adj. P-Value: (127. T3 Tc WT) / (126. T0 WT)': float,
        '-LOG10 Adj.P-val': float
    },
    na_values="#VALEUR!"  # treat "#VALEUR!" as missing
)
```

```python
fig, ax = plt.subplots()
ax.hist(df['Log2 Corrected Abundance Ratio'])
plt.show()
```

```python
data = df['Log2 Corrected Abundance Ratio']  # select numeric column
mu = data.mean()     
sigma = data.std()   

print(mu, sigma)    
```

```python
# abundance vector
data = df['Log2 Corrected Abundance Ratio'] 

# Estimate mean and std
mu = np.mean(data)
sigma = np.std(data, ddof=1)  # sample std

# Create histogram
fig, ax = plt.subplots()
hist = ax.hist(data, bins=100, color='skyblue', edgecolor='black', alpha=0.7)

# PDF domain points
x = np.linspace(min(data), max(data), 100)

# Scaling factor for PDF to match histogram
dx = hist[1][1] - hist[1][0]   # bin width
scale = len(data) * dx          # scale PDF

# Plot PDF on top
ax.plot(x, norm.pdf(x, mu, sigma)*scale, color='red', linewidth=2)

# Labels
ax.set_xlabel('Log2 Corrected Abundance Ratio')
ax.set_ylabel('Frequency')
ax.set_title('Histogram with Normal Distribution Overlay')
plt.savefig("data/log2_histogram.png", dpi=150)
plt.show()
```
